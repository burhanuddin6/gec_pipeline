```python
import os
import xml.etree.ElementTree as ET
# This is the correct import for stable versions of UrduHack
from urduhack.tokenization import sentence_tokenizer
from tqdm import tqdm
import urduhack

def save_sentences_to_txt(corpus_path, output_file_path):
    """
    Extracts all sentences from the Maḵẖzan corpus and saves them to a text file.

    Args:
        corpus_path (str): The path to the cloned Maḵẖzan repository.
        output_file_path (str): The path where the output .txt file will be saved.
    """
    text_directory = os.path.join(corpus_path, 'text')
    if not os.path.isdir(text_directory):
        print(f"Error: The directory '{text_directory}' was not found.")
        return

    # A list to hold all sentences from the entire corpus
    all_sentences = []

    xml_files = [os.path.join(root, file)
                 for root, _, files in os.walk(text_directory)
                 for file in files if file.endswith('.xml')]

    print(f"Found {len(xml_files)} XML files to process...")

    # Process each XML file and collect sentences
    for file_path in tqdm(xml_files, desc="Extracting sentences"):
        try:
            tree = ET.parse(file_path)
            root = tree.getroot()

            for paragraph in root.iter('p'):
                if paragraph.text:
                    paragraph_text = paragraph.text.strip()
                    if paragraph_text:
                        # Tokenize the paragraph into sentences
                        sentences = sentence_tokenizer(paragraph_text)
                        # Add the sentences from this paragraph to our main list
                        all_sentences.extend(sentences)
        except ET.ParseError as e:
            print(f"Could not parse file {os.path.basename(file_path)}: {e}")
        except Exception as e:
            print(f"An error occurred with file {os.path.basename(file_path)}: {e}")

    # Check if we actually found any sentences
    if not all_sentences:
        print("No sentences were extracted. The output file will be empty.")
        return

    # Write all collected sentences to the output file
    print(f"\nWriting {len(all_sentences):,} sentences to {output_file_path}...")
    try:
        with open(output_file_path, 'w', encoding='utf-8') as f:
            # Join all sentences with a newline character and write to the file
            f.write('\n'.join(all_sentences))
        print("Successfully saved the sentences.")
    except Exception as e:
        print(f"An error occurred while writing to the file: {e}")


if __name__ == "__main__":
    # Path to the cloned makhzan repository
    makhzan_repo_path = './makhzan'
    # Name of the file where sentences will be saved
    output_filename = 'makhzan_sentences.txt'

    print("Starting the sentence extraction process for the Maḵẖzan corpus...")

    # Ensure UrduHack models are downloaded
    try:
        urduhack.download()
    except Exception as e:
        print(f"Could not download UrduHack models: {e}")
        exit()

    save_sentences_to_txt(makhzan_repo_path, output_filename)

    print("\n-----------------------------------------")
    print(f"Process finished. The corpus of sentences is saved in '{output_filename}'.")
    print("-----------------------------------------")
```
