```python
import shutil

def create_replica(source_file, replica_file):
  try:
    shutil.copy(source_file, replica_file)
    print(f"Successfully created replica: {replica_file}")
  except FileNotFoundError:
    print(f"Error: The source file '{source_file}' was not found.")
  except Exception as e:
    print(f"An error occurred: {e}")

# Example usage:
create_replica("wikiedits_urdu/urwiki.extracted.edits", "data/raw_wikiedits.txt")
```
